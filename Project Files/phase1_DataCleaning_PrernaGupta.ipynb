# Cheking for missing values in every column for every file

```python
def checking_missing_values(df, name="data_frame"):
    print(f'{'-' * 50} {name} {'-' * 50}')
    # getting the columns of the dataframe
    cols = list(df.columns)
    null_rows = False

    for col in cols:
        null_values = df[col].isnull().sum()
        if null_values > 0:
            null_rows = True
            percentage_rows_affected = round((null_values / len(df)) * 100.00, 2)
            print(f'{col} in {name} has {null_values} null values which affects {percentage_rows_affected}% of rows.')
    
    if null_rows is False:
        print(f'{name} has 0 null values.')

    
```

```python
checking_missing_values(ratings_df, "ratings.csv")
```

# Checking for duplicate rows

```python
def checking_duplicate_rows(df, name="data_frame"):
    print(f'{'-' * 50} {name} {'-' * 50}')
    # getting the columns of the dataframe
    cols = list(df.columns)

    contain_duplicate = False

    for col in cols:

        duplicate_rows = df[col].duplicated().sum()

        if duplicate_rows > 0:
            contain_duplicate = True
            print(f'{col} in {name} has {duplicate_rows} duplicate rows.')
    
    if contain_duplicate is False:
        print(f'{name} does not have any columns which have duplicate rows.')
```

# Checking for outliers in watch_duration_min

```python
import numpy as np

watch_duration = watch_history_df['watch_duration_min']
mean = np.mean(watch_duration)
std = np.std(watch_duration)

lower_bound = mean - 4 * std
upper_bound = mean + 4 * std

# Enclose each boolean condition in parentheses
statistical_outliers = (watch_history_df['watch_duration_min'] < lower_bound) | (watch_history_df['watch_duration_min'] > upper_bound)
logical_violations = (watch_history_df['watch_duration_min'] > watch_history_df['content_duration_min'])

outliers_watch_history = watch_history_df[statistical_outliers | logical_violations]

# Create specific flag columns
watch_history_df['is_stat_outlier'] = (watch_history_df['watch_duration_min'] < lower_bound) | (watch_history_df['watch_duration_min'] > upper_bound)
watch_history_df['exceeds_content_len'] = watch_history_df['watch_duration_min'] > watch_history_df['content_duration_min']

# Combined flag
watch_history_df['is_flagged'] = watch_history_df['is_stat_outlier'] | watch_history_df['exceeds_content_len']

# Filter flagged records
outliers_watch_history = watch_history_df[watch_history_df['is_flagged']]
print(f"Total flagged sessions: {len(outliers_watch_history)}")

print(watch_history_df[watch_history_df['is_flagged'] == True])
```

# Verifying Referential Integrity

```python
def missing_fk(child_df, parent_df, fk_col, pk_col=None, df_name="Child Table"):
    """Identifies foreign key records in child_df that do not exist in parent_df."""
    if pk_col is None:
        pk_col = fk_col

    # Identify records in the child table missing from the parent table
    orphan_records = child_df[~child_df[fk_col].isin(parent_df[pk_col])]

    orphan_unique_count = orphan_records[fk_col].nunique()
    total_orphans = len(orphan_records)

    print(f"--- Foreign Key Integrity Check: {df_name} ---")
    print(f"Total orphan records: {total_orphans}")
    print(f"Unique missing '{fk_col}' keys: {orphan_unique_count}")

    if total_orphans > 0:
        print("\nSample orphan records:")
        print(orphan_records.head())

    return orphan_records
```

## Detailed Data Quality & Integrity Findings

### 1. Primary Key & Duplicate Analysis
* **Primary Keys:** All primary keys (`rating_id`, `review_id`, `subscriber_id`, `title_id`, `watch_id`, `watchlist_id`) are 100% unique, verifying that entity constraints are strictly met.
* **Non-Key Cardinality & Duplication:** 
  * High duplicate counts across user/content interaction keys confirm expected one-to-many relationships (e.g., `subscriber_id` in `watch_history`: 635,002 duplicates; `title_id` in `ratings`: 121,001 duplicates).
  * **`titles.csv` Title Name Duplicates:** 53 duplicate entries exist in `title_name` (representing multi-season entries, re-releases, or localized variations sharing identical names).

### 2. Missing Value Profiling
* Structural missingness is confined strictly to two temporal attributes dependent on business logic:
  * **`subscribers.csv` (`churn_date`):** **74.66% missing** (11,199 records). These nulls represent active subscribers who have not churned (supported by `is_active == True`).
  * **`titles.csv` (`license_expiry`):** **21.84% missing** (1,966 records). These represent owned/original content or perpetual licenses with no expiration date.
* All other tables (`ratings`, `reviews`, `watch_history`, `watchlist`) contain **zero null values** across all columns.

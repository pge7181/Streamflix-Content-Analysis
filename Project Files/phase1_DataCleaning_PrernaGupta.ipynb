# Running Basic Profile Test for all the files

```python
def runnning_basic_profile(df, name="data_drame"):
    print(f'{'-' * 50} {name} {'-' * 50}')
    # getting the columns of the dataframe
    cols = list(df.columns)

    # printing the number of rows in csv file
    print(f'The row count in {name} is: {len(df)}')
    print(f'{'*' * 100}')
    # printing the number of columns in csv file
    print(f'The number of columns in {name} is: {len(cols)}')
    print(f'{'*' * 100}')
    # printing the datatypes for every column in the df
    for col in cols:

        print(f'The data type for {col} column in {name} is: {df[col].dtype}')
```

# Cheking for missing values in every column for every file

```python
def checking_missing_values(df, name="data_frame"):
    print(f'{'-' * 50} {name} {'-' * 50}')
    # getting the columns of the dataframe
    cols = list(df.columns)
    null_rows = False

    for col in cols:
        null_values = df[col].isnull().sum()
        if null_values > 0:
            null_rows = True
            percentage_rows_affected = round((null_values / len(df)) * 100.00, 2)
            print(f'{col} in {name} has {null_values} null values which affects {percentage_rows_affected}% of rows.')
    
    if null_rows is False:
        print(f'{name} has 0 null values.')

    
```

```python
checking_missing_values(ratings_df, "ratings.csv")
```

```python
checking_missing_values(reviews_df, "reviews.csv")
```

```python
checking_missing_values(subscribers_df, "subscribers.csv")
```

```python
checking_missing_values(titles_df, "titles.csv")
```

```python
checking_missing_values(watch_history_df, "watch_history.csv")
```

```python
checking_missing_values(watchlist_df, "watchlist.csv")
```

# Checking for duplicate rows

```python
def checking_duplicate_rows(df, name="data_frame"):
    print(f'{'-' * 50} {name} {'-' * 50}')
    # getting the columns of the dataframe
    cols = list(df.columns)

    contain_duplicate = False

    for col in cols:

        duplicate_rows = df[col].duplicated().sum()

        if duplicate_rows > 0:
            contain_duplicate = True
            print(f'{col} in {name} has {duplicate_rows} duplicate rows.')
    
    if contain_duplicate is False:
        print(f'{name} does not have any columns which have duplicate rows.')
```

```python
checking_duplicate_rows(ratings_df, "ratings.csv")
```

```python
checking_duplicate_rows(reviews_df, "reviews.csv")
```

```python
checking_duplicate_rows(subscribers_df, "subscribers.csv")
```

```python
checking_duplicate_rows(titles_df, "titles.csv")
```

```python
checking_duplicate_rows(watch_history_df, "watch_history.csv")
```

```python
checking_duplicate_rows(watchlist_df, "watchlist.csv")
```
